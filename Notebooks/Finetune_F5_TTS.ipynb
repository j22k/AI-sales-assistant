```python
from huggingface_hub import hf_hub_download
from pathlib import Path
import shutil

MODEL_ID = "multilingual-tts/F5-TTS-OpenBible-Malayalam"

CHECKPOINT_DIR = PROJECT_DIR / "checkpoints"
CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

hf_checkpoint = hf_hub_download(
    repo_id=MODEL_ID,
    filename="model_last.pt"
)

PERSISTENT_CHECKPOINT = CHECKPOINT_DIR / "model_last.pt"

shutil.copy2(hf_checkpoint, PERSISTENT_CHECKPOINT)

print("Downloaded checkpoint:", hf_checkpoint)
print("Persistent checkpoint:", PERSISTENT_CHECKPOINT)
print("Exists:", PERSISTENT_CHECKPOINT.exists())
print("Size:", round(PERSISTENT_CHECKPOINT.stat().st_size / (1024**3), 2), "GB")
```

```python
import torch

CHECKPOINT_PATH = PROJECT_DIR / "checkpoints" / "model_last.pt"

print("Checkpoint:", CHECKPOINT_PATH)
print("Exists:", CHECKPOINT_PATH.exists())
print("Size:", round(CHECKPOINT_PATH.stat().st_size / (1024**3), 2), "GB")

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location="cpu",
    weights_only=False
)

print("\nCheckpoint type:", type(checkpoint))

if isinstance(checkpoint, dict):
    print("Checkpoint keys:")
    for key in checkpoint.keys():
        print(" -", key)
```

```python
state = checkpoint["model_state_dict"]

matches = []

for key, value in state.items():
    if "text_embed" in key and hasattr(value, "shape"):
        matches.append((key, tuple(value.shape)))

print("Text embedding tensors:")
for key, shape in matches:
    print(f"{key}: {shape}")

print("\nVocabulary size:", 119)
```

```python
from f5_tts.model.utils import get_tokenizer

vocab_char_map, vocab_size = get_tokenizer(
    str(VOCAB_PATH),
    "custom"
)

print("Tokenizer-reported vocab size:", vocab_size)
print("Number of vocabulary entries :", len(vocab_char_map))

print("\nLast 10 vocabulary entries:")
for token, index in list(vocab_char_map.items())[-10:]:
    print(repr(token), "->", index)

print("\nHighest token index:", max(vocab_char_map.values()))
```

```python
from f5_tts.model.backbones.dit import DiT
from f5_tts.model import CFM

MODEL_CONFIG = {
    "dim": 1024,
    "depth": 22,
    "heads": 16,
    "ff_mult": 2,
    "text_dim": 512,
    "text_mask_padding": True,
    "qk_norm": None,
    "conv_layers": 4,
    "pe_attn_head": None,
    "attn_backend": "torch",
    "attn_mask_enabled": False,
    "checkpoint_activations": False,
}

model = CFM(
    transformer=DiT(
        **MODEL_CONFIG,
        text_num_embeds=119,
        mel_dim=100,
    ),
    mel_spec_kwargs={
        "target_sample_rate": 24000,
        "n_mel_channels": 100,
        "hop_length": 256,
        "win_length": 1024,
        "n_fft": 1024,
        "mel_spec_type": "vocos",
    },
    vocab_char_map=vocab_char_map,
)

model.load_state_dict(
    checkpoint["model_state_dict"],
    strict=True,
)

print("Model + checkpoint compatibility: OK")
```

```python
from pathlib import Path

PROJECT_DIR = Path(
    "/content/drive/MyDrive/Datsets/F5-TTS-Malayalam"
)

PREPARED_DATASET = (
    PROJECT_DIR / "dataset" / "kathbath_malayalam_finetune"
)

VOCAB_PATH = (
    PROJECT_DIR / "dataset" / "vocab.txt"
)

CHECKPOINT_PATH = (
    PROJECT_DIR / "checkpoints" / "model_last.pt"
)

PATCHED_SCRIPT = (
    PROJECT_DIR / "scripts" / "finetune_malayalam.py"
)

# Verify everything before constructing the command
for name, path in {
    "PREPARED_DATASET": PREPARED_DATASET,
    "VOCAB_PATH": VOCAB_PATH,
    "CHECKPOINT_PATH": CHECKPOINT_PATH,
    "PATCHED_SCRIPT": PATCHED_SCRIPT,
}.items():
    print(f"{name}: {path}")
    print(f"  exists = {path.exists()}")

assert PREPARED_DATASET.exists()
assert VOCAB_PATH.exists()
assert CHECKPOINT_PATH.exists()
assert PATCHED_SCRIPT.exists()

print("\nAll required paths exist.")
```

```python
accelerate_cmd = [
    "accelerate",
    "launch",

    "--mixed_precision", "fp16",
    "--num_processes", "1",

    str(PATCHED_SCRIPT),

    "--exp_name", "F5TTS_v1_Base",
    "--dataset_name", str(PREPARED_DATASET),

    "--learning_rate", "1e-5",

    # Conservative T4 test configuration
    "--batch_size_per_gpu", "800",
    "--batch_size_type", "frame",
    "--max_samples", "4",
    "--grad_accumulation_steps", "1",
    "--max_grad_norm", "1.0",

    "--epochs", "1",
    "--num_warmup_updates", "10",

    "--save_per_updates", "100",
    "--keep_last_n_checkpoints", "1",
    "--last_per_updates", "50",

    "--finetune",
    "--pretrain", str(CHECKPOINT_PATH),

    "--tokenizer", "custom",
    "--tokenizer_path", str(VOCAB_PATH),

    "--logger", "tensorboard",
]

print("\n========== COMMAND ==========")
print(" ".join(accelerate_cmd))
```
